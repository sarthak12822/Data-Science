```python
import matplotlib.pyplot as plt
import numpy as np
x = [1,2,3,4,5]
y = [10,20,15,30,25]
```

```python
plt.plot (x,y,color = "green", linewidth = 1.5, linestyle = "-.", marker = "*")
#linestyle for style of line(-. , - , : , --), linewidth is for the width of line
#marker = "o","s","^","*","+"."x"
plt.title ("Sundar graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid()
plt.show()
```

```python
month = [1,2,3,4,5]
sales_A =[10,20,15,30,25]
sales_B= [25,30,35,40,55]
plt.plot (month,sales_A,color = "green", linewidth = 1.5, linestyle = "-.", marker = "*")
plt.plot (month,sales_B)
plt.show()
```

```python
#pi chart
#percentage = (category value /total)*100
#(percentage/100)*360 = degrees
import matplotlib.pyplot as plt
import numpy as np
dept =["mech","civil","rai","cse"]
students = [60,70,120,200]
color = ["purple","pink","brown","orange"]
plt.pie (students,labels=dept,autopct="%1.1f%%",colors = color,startangle=90)
plt.title("Department")
plt.show()
```

```python
#box plot
#Quartiles,medians,IQR,whiskers,outlies,will draw the box plot
#box plot is statsical graph used to summerise the distribution of numerical data usig 5 importance values(five number summery)
#minimum ,Q1= mediam of the lower half,median (Q2),(Q3)=median of the upper half,Maximum,
#IQR (interquartile rang ) =Q3-Q1
#outliers
#lowelimit = Q1 - 1.5 (IQR)
#upperlimt = Q3 + 1.5 (IQR)
import numpy as np
import matplotlib.pyplot as plt
marks =[10,15,20,22,24,26,28,30,35,95]
percentage= [75,41,25,13,15,15,18,85,85,95]
plt.boxplot (marks,patch_artist=True,boxprops=dict (facecolor = "red"),showmeans=True)
plt.boxplot (percentage ,patch_artist=True,boxprops=dict (facecolor = "red"),showmeans=True)
plt.show()
```

```python
#Subplot
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,4,6,8,10]
marks = [45,50,55,60,65]
plt.figure (figsize = (10,8))
plt.subplot(2,2,1)
plt.plot(x,y)
plt.title("Line Plot")
plt.subplot(2,2,2)
plt.hist(marks)
plt.title("Histogram")
plt.subplot(2,2,3)
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.subplot(2,2,4)
plt.bar(x,y)
plt.title("BAR PLOT")

plt.show()
```

```python
#Advance customization techniques
#Fig size
#Axis Limit
#Subplot   #multiple styles ==> ggplot,classic,bmh
import  matplotlib.pyplot as plt
import numpy as np
x = [1,2,3,4,5]
y = [2,4,6,8,10]
marks = [45,50,55,60,65,70,75,80]
plt.figure (figsize = (10,8))
plt.figure (facecolor= "green")
plt.plot (x,y)
plt.xlim (0,10)
plt.xticks([1,2,3,4,5],["Jan","Feb","Mar","Apr","May"])
plt.title ("monthly sales",fontsize = 16)
plt.xlabel ("Months",fontsize = 14)
plt.ylabel ("Sales",fontsize = 14)
plt.grid ()
plt.ylim (0,15)
plt.savefig("Monthly sales.png",dpi = 300)
plt.savefig ("Graph.pdf")
plt.show()
```
